```python
#3
n = int(input("Enter a number: "))
if n < 2: print( str(n) + " is not a prime number ")
elif n == 2: print(str(n) + " is a prime number ")
else:
    for i in range(2, n):
        if n % i == 0:
            print(str(n) + " is not a prime number ")
            break
        else: print(str(n) + " is a prime number ")
        break
```

```python
#4
n = int(input("Enter a number: "))
def number(n):
    numb = []
    for i in range(1, n):
      if n % i == 0:
        numb.append(i)
    return numb


total = sum(number(n))
if total == n:
  print(str(n) + " is a perfect number ")
else: print(str(n) + " is not a perfect number ")
```

```python
#7
def remove_dollar_sign(s):
  new_str = ""                       # return "".join[c for c in s if c != ]
  for char in s:
    if char != "$":
      new_str += char
  return new_str
input_str = input("Enter a string: ")
result = remove_dollar_sign(input_str)
print(result)
```

```python
#8
def extract_even():
  list1 = [1,4,5,-1,10]
  new_list = []
  for i in list1:
    if i % 2 == 0:
      new_list.append(i) # =       new.list += [i]
  return new_list
result = extract_even()
print(result)
```

```python
#9
n = int(input("Enter a number: "))
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)
print(factorial(n))
```

```python
#10
n = int(input("Enter a number: "))
def divisor_numb(n):
  divisor_list = []
  for i in range(1, n+1):
    if n % i == 0:
      divisor_list.append(i)
  return divisor_list
print(divisor_numb(n))
```

```python
#11
import math

def dist_2_p(x, y, a, b):
  distance =  math.sqrt((a-x)**2 + (b-y)**2)
  return distance
x = int(input("Enter x1: "))
y = int(input("Enter y1: "))
a = int(input("Enter x2: "))
b = int(input("Enter y2: "))
print(dist_2_p(x, y, a, b))
```

```python
#12
def print_rectangle(m, n):
    for i in range(m):
        for j in range(n):
            if i == 0 or i == m-1 or j == 0 or j == n-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()
print_rectangle(4, 5)

```
